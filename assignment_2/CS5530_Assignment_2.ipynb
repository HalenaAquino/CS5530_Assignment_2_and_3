```python
# load the data
from pathlib import Path
import pandas as pd
import numpy as np

local_path = Path("data/train.csv")
data_url = 'https://app.box.com/s/jm6pw202asu4xd3uypwtry2rqk691y1i'

source = local_path if local_path.exists() else data_url
df_raw = pd.read_csv(source)

print(f'Loaded {df_raw.shape[0]} rows and {df_raw.shape[1]} columns from {source}')
```

```python
# Task a (handle missing values)
df_raw.isna().sum()

df_cols = df_raw.columns.tolist()
df_cols.remove('New_Price')

df_clean = df_raw.dropna(subset=df_cols).reset_index(drop=True)

df_clean.head()
```

Because of the size of the data (> 5000) rows, I decided to just remove the rows with any missing data. However, I didn't drop the rows with a missing New_Price value because minimal rows actually have a value for that column. Essentially, any rows with a missing value for any row except New_Price were dropped

```python
# Task b (remove units)
df_clean = df_clean.copy()
df_clean['Engine'] = (df_clean['Engine'].str.replace('CC', ''))
df_clean['New_Price'] = (df_clean['New_Price'].str.replace('Lakh', ''))
df_clean['Power'] = (df_clean['Power'].str.replace('bhp', ''))
df_clean['Mileage'] = (df_clean['Mileage'].str.replace('kmpl', '').str.replace('km/kg', ''))

df_clean.head()
```

```python
# Task c (change categorical into numerical)
df_clean = pd.get_dummies(df_clean, columns=['Fuel_Type', 'Transmission'])
df_clean.head()
```

```python
# Task d (create new feature)
car_ages = []

for i in range(len(df_clean)):
  car_ages.append(2026 - int(df_clean['Year'][i]))

df_clean.insert(4, 'Car Age', car_ages)

df_clean.head()
```

```python
# Task e (descriptive operations)

# select
selected_df = df_clean[['Name', 'Year', 'Car Age']]

selected_df.head()
```

```python
# filter
filtered_df = df_clean.loc[df_clean['Car Age'] > 10]

filtered_df.head()
```

```python
# rename
renamed_df = df_clean.rename(columns={'Kilometers_Driven': 'Km_Driven'})

renamed_df.head()
```

```python
# mutate
mutated_df = renamed_df.assign(Km_Driven_Per_Year = renamed_df['Km_Driven'] / renamed_df['Car Age'])

mutated_df.head()
```

```python
# arrange
arranged_df = mutated_df.sort_values(by=['Km_Driven_Per_Year'], ascending=False)

arranged_df.head()
```

```python
# summarize
summary_df = (arranged_df
              .groupby('Car Age')
              .agg(avg_km_driven_per_year=('Km_Driven_Per_Year', 'mean'),
                   median_km_driven_per_year=('Km_Driven_Per_Year', 'median'),
                   num_cars=('Name', 'count'))
              .reset_index())

summary_df.head()
```
